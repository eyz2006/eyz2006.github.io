## Step 1

Print out the trip distances.


```python
# AI code here
import pandas as pd

# Load the CSV file
df = pd.read_csv("yellow_taxi.csv")

# Display the first few rows to understand the structure
print(df.head())

# Display all column names to find the one for distance
print(df.columns)
```

```python
# Print out the distance column
print(df["trip_distance"])
```

## Step 2

Calculate the average ride distance.


```python
# AI code here
# Calculate the average ride distance
average_distance = df["trip_distance"].mean()

print("Average ride distance:", average_distance)
```

## Step 3

Your turn! Calculate the percent of trips that were paid for by credit card. The data dictionary will be helpful - see the Attachment on [the dataset page](https://data.cityofnewyork.us/Transportation/2023-Yellow-Taxi-Trip-Data/4b4i-vvec/about_data).


```python
# AI code here
# Calculate percentage of trips paid by credit card
percent_credit_card = (df["payment_type"].eq(1).mean()) * 100

print(f"Percent of trips paid by credit card: {percent_credit_card:.2f}%")
```

## Step 4

Save a random sample of the trips to a new CSV.


```python
# AI code here
# Take a random sample — for example, 1000 rows (you can change this number)
sample_df = df.sample(n=1000, random_state=42)

# Save the sample to a new CSV file
sample_df.to_csv("yellow_taxi_sample.csv", index=False)

print("✅ Random sample saved as 'yellow_taxi_sample.csv'")
```
